```python
import numpy as np
def euclidean(vec1,vec2):
 distance=np.array(vec1)-np.array(vec2)
 squared_sum=np.sum(distance**2)
 return np.sqrt(squared_sum)
```

```python
sword = [10,1,5,1]
knife = [10,1,5,9]
herring = [2,2,2,6]

print(euclidean(sword,knife))
print(euclidean(sword,herring))
print(euclidean(knife,herring))
```

```python
def dot_product(vector1,vector2):
    thedotproduct=np.sum([vector1[k]*vector2[k] for k in range(0,len(vector1))])
    return(thedotproduct)
def vector_norm(vector):
    thenorm=np.sqrt(dot_product(vector,vector))
    return(thenorm)
def cosine_similarity(vector1,vector2):
    thecosine=0
    thedotproduct=dot_product(vector1,vector2)
    thecosine=thedotproduct/(vector_norm(vector1)*vector_norm(vector2))
    thecosine=np.round(thecosine,4)
    return(thecosine)
```

```python
print(cosine_similarity(vectors['sword'],vectors['knife']))
print(cosine_similarity(vectors['sword'],vectors['herring']))
print(cosine_similarity(vectors['car'],vectors['van']))
```

```python
king = vectors['king']
queen = vectors['queen']
man = vectors['man']
woman = vectors['woman']

newvector = king-man+woman

print(cosine_similarity(newvector,queen))
print(euclidean(newvector,queen))
```

```python
Sentences = [
    "The trouble with having an open mind, of course, is that people will insist on coming along and trying to put things in it.",\
    "The problem with having an open mind is that people will insist on approaching and trying to insert things into your mind.",\
    "To be or not to be, that is the question",\
    "Call me Ishmael"
]
```

```python
import tensorflow_hub as hub
embed = hub.load("https://tfhub.dev/google/universal-sentence-encoder-large/5")

embeddings = embed(Sentences)

print(embeddings[0])
```

```python
print(cosine_similarity(embeddings[0],embeddings[1]))
print(cosine_similarity(embeddings[0],embeddings[2]))
print(cosine_similarity(embeddings[0],embeddings[3]))
```

```python
Sentences = [
 "The corn and cheese are delicious when they're roasted together",
 "Several of the scenes have rich settings but weak characterization",
 "Consider adding extra seasoning to the pork",
 "The prose was overwrought and pretentious",
 "There are some nice brisket slices on the menu",
 "It would be better to have a chapter to introduce your main plot ideas",
 "Everything was cold when the waiter brought it to the table",
 "You can probably find it at a cheaper price in bookstores"
]

embeddings = embed(Sentences)
```

```python
arrays=[]

for i in range(len(Sentences)):
 arrays.append(np.array(embeddings[i]))

sentencematrix = np.empty((len(Sentences),512,), order = "F")
for i in range(len(Sentences)):
 sentencematrix[i]=arrays[i]

import pandas as pd
pandasmatrix=pd.DataFrame(sentencematrix)
```

```python
from sklearn.cluster import KMeans
m = KMeans(2)
m.fit(pandasmatrix)
pandasmatrix['topic'] = m.labels_
pandasmatrix['sentences']=Sentences
```

```python
print(pandasmatrix.loc[pandasmatrix['topic']==0,'sentences'])
print(pandasmatrix.loc[pandasmatrix['topic']==1,'sentences'])
```

Ch 10 Summary:

This chapter deals with something called natural language processing (NLP). This is an entire field dedicated to converting text into mathematical equivalents that we can use to analyze and compare.

The scenario we spend most of the chapter on is plagiarism detection. Using an NPL tool can help detect close similarities between texts that are not exact copies of each other (paraphrasing instead of copy pasting). Humans can easily intuit the similarities between sentences or words with our natural understanding of synonyms and denotation. If we want to have a computer mimic this, we have to use something called a natural language corpus that includes normal human speech patterns and recognizes word similarity.

The natural language corpus can provide us with the probability of two words appearing next to each other, based off of their definitions. This percentage forms the basis of our mathematical analysis. We then take our percentages and rewrite them in a system of equations with the probability equalling the dot product of two vectors.

Now that we have the mathematical understanding of how to denote the similarities or proximities between words, we can write a Python program to accomplish this work. The function euclidean(vec1, vec2) calculates the distance between the two given word vectors. The greater the given distance, the less similar the two words are. We can also understand the probability of words appearing next to each other as a neural network. So, we import the word2vec model to create our word vectors. The word2vec model contains about 100 billion words and is based off of actual, human-written text. We can use it to get a vector for any given word and then use our euclidean function to get our distances much faster than before. Cosine similarity, that we used in the previous chapter, can also be used to calculate differences between words.

To experiment with the new model and our calculating function, the chapter takes us through the example of seeing if calculating the vector equation king - man + woman will give us a similar vector to the one for queen. Ultimately, after using both the euclidean and cosine_similarity function, that the vector similarities are fairly close.

Now, the chapter returns to the plagiarism scenario. The first problem we run into is that we only know how to compare individual words, not entire sentences. It would take a long time to work through an entire essay, for example, if we were just comparing it word by word. Instead, we load in something called the universal sentence encoder (USE) through the tensorflow_hub module. Running sentences through the USE model, saved as the variable “embeddings”, we get a vector with hundreds of elements. We then can use our cosine_similarity function to check the vectors of sentences.

Now, we move on to something called topic modeling which is a method of gathering main ideas from a collection of text. Our example text includes sentences that discuss food and books. We first calculate the vectors for the sentences using the embeddings variable and embed function. Next, we use two different for loops to add the the sentence embeddings to an array matrix. The text then shows us how to incorporate k means clustering to find those main ideas from the sentence vectors. The result is that our group of sentences are split into two clusters based off their meanings: one for food and one for books.

The chapter ends with a discussion of some other applications of natural language processing. NLP can be used in content-based recommendation systems, sentiment analysis (assigning tone to text), and basic chatbots.
