# Problem 1

```python
def hanoi(p, source, target, intermediate):
  if p==1:
    print("move disk from", source, "to", target)
  else:
    hanoi(p-1, source, intermediate, target)
    hanoi(1, source, target, intermediate)
    hanoi(p-1, intermediate, target, source)

hanoi(4, "left", "middle", "right")
```

# Problem 3

```python
import re

def pigLatin(input):
  message = input

  vowel_regex = re.compile(r'[aeiouAEIOU]')
  consonant_regex = re.compile(r'[^aeiouAEIOU]')

  base = list(message)
  if not (vowel_regex.search(base[0])== None):
    print(message+'way')
  else:
    if (vowel_regex.search(base[1])== None):
      print("".join(base[2:]) + "".join(base[0:2]) + 'ay')
    else:
      print("".join(base[1:]) + base[0] + 'ay')

print("Enter the sentence to translate: ")
sentence = input()
statement = []

for word in sentence.split():
  statement.append(pigLatin(word))
print(statement)
```

```python
import re

x = input()
words = x.split()
transformed_words = []

for word in words:
  word = re.sub(r'\s', '', word)
  if re.match(r'^[BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz][AEIOUaeiou]\w*$', word):
    word = re.sub(r'^([BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz])([AEIOUaeiou])(\W*)(\W*)$', r'\2\3\1ay\4', word)
  elif re.match(r'^[BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz][BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz]\w*W*$', word):
    word = re.sub(r'^[BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz])([BCDFGHJKLMNPQRSTVWXYZbcdfghjklmnpqrstvwxyz])', r'\3\1\2ay\4', word)
  elif re.sub(r'^[AEIOUaeiou]\w*W*$', word):
    word = re.sub(r'^([AEIOUaeiou])(\w*)(\W*)$', r'\1\2way\3', word)
  transformed_words.append(word)

print(" ".join(transformed_words))
```
