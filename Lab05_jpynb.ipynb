Q24

```python
## Dependencies
import numpy as np
## Data: yearly salary in ($1000) [2017, 2018, 2019]
alice = [99, 101, 103]
bob = [110, 108, 105]
tim = [90, 88, 85]
names = ['alice', 'bob', 'tim']
salaries = np.array([alice, bob, tim])
taxation = np.array([[0.2, 0.25, 0.22],
 [0.4, 0.5, 0.5],
 [0.1, 0.2, 0.1]])
## One-liner
max_income = np.max(salaries - salaries * taxation)

## Result
print(max_income)

#  pull the name of the person with the highest salary from the matrix by using this idea of selective Boolean indexing. Problem recap: How do we find the person with maximum after-tax income in a group of  people, given their yearly salary and tax rates?
after_tax_incomes = (salaries - salaries * taxation).max(axis=1)
print(after_tax_incomes)
max_index = np.argmax(after_tax_incomes)
print(max_index)
print(names[max_index])
```

Q25

```python
a = np.array([[[1, 2], [3, 4]],
              [[5, 6], [7, 8]]])
print(np.average(a, axis=0))
print(np.average(a, axis=1))
print(np.average(a, axis=2))
```

The three different print statements result in the average of the array's contents across its different axes. The first statement averages the values vertically in their current shape (ex: the average of 1 and 5 is 3 which is why 3 is the first value printed).
The second print statement averages the values across the first entry of each of the 1x1 arrays (ex: the average of 1 and 3 is 2 which is why 2 is the first value printed).
The third print statement averages the values within each 1x1 array in a (ex: the average of 1 and 2 is 1.5 which is why 1.5 is what is printed first).

Q26

```python
a = np.array([[[ 1,  2,  3], [22,  5,  6], [ 7,  8,  9]],
              [[11, 12, 10], [ 4, 14, 15], [16, 17, 18]],
              [[19, 20, 21], [13, 23, 24], [25, 26, 27]]])

b = np.sort(a, axis=0)
print(b)
print()
c = np.sort(a, axis=1)
print(c)
print()
d = np.sort(a, axis=2)
print(d)
```

The first sort function arranges the values of the first column of array in ascending numerical order. This is why 4, 22, and 13 all switch places.
The second sort function rearranged the 1x3 arrays to make sure that the first values of each array are in order. That is why 7 and 22 switched places, for example, because 1, 7, and 22 had to be placed in ascending order.
The third sort function arranges the values within the 1x3 arrays to be in ascending order. For example, [22, 5, 6] switches to [5, 6, 22].
