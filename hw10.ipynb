```python
import pandas as pd
desktop=pd.read_csv('desktop.csv')
laptop=pd.read_csv('laptop.csv')
```

```python
print(desktop.head())
print(laptop.head())
```

```python
import scipy.stats
print(scipy.stats.ttest_ind(desktop['spending'],laptop['spending']))
print(scipy.stats.ttest_ind(desktop['age'],laptop['age']))
print(scipy.stats.ttest_ind(desktop['visits'],laptop['visits']))
```

```python
import numpy as np
medianage=np.median(desktop['age'])
groupa=desktop.loc[desktop['age']<=medianage,:]
groupb=desktop.loc[desktop['age']>medianage,:]
```

```python
emailresults1=pd.read_csv('emailresults1.csv')
print(emailresults1.head())
```

```python
groupa_withrevenue=groupa.merge(emailresults1,on='userid')
groupb_withrevenue=groupb.merge(emailresults1,on='userid')
print(scipy.stats.ttest_ind(groupa_withrevenue['revenue'],groupb_withrevenue['revenue']))
```

```python
print(np.mean(groupb_withrevenue['revenue'])-np.mean(groupa_withrevenue['revenue']))
```

Ch 4 A/B Testing:

This chapter begins with the same scenario from the second half of the previous chapter--we are trying to determine the differences between the selling of laptops and desktops to better inform our business's marketing strategies.

We start off by using t-tests to see if there are statistically significant differences between laptops and desktops in three difference categories: spending, age, and visits. Spending is the only category where we find a statistically significant difference for laptop and desktop buyers.

The book proceeds to explain the A/B testing process in great detail. It consists of "experimental design, random assignment into treatment and control groups, careful measurement of outcomes, and finally,
statistical comparison of outcomes between groups" (page 77).

First, we wanted to find if changing the color of the text in the marketing emails would affect revenue. We focus just on the desktop group by creating a group A and group B out of the sample. We conduct the t-test after merging the groups with their results from the email campaign results. We then calculate the difference, $125, between the means of the two group. Therefore, the average member of group B outspends the average member of group A by $125. This is a significant change in spending habits.

However, the textbook reveals that we cannot trust that result because there is a confounding variable impacting the comparison. There is an additional difference between group A and group B besides the different text colors on the email.

The goal of A/B testing is to find how the expected value of outcome A and the expected value of factor B contribute to the expected value of outcome C. So for the new attempt, we will create truly random samples, Groups C and D. After connecting C and D to their email results, we find that there is a statistically significant difference of ~$260 in revenue between the two groups.

The text recommends that A/B testing be a continuous process so that your product can keep improving rather than stagnate. We should keep pitting new models against the current option in A/B testing and keep replacing the current model with the winners. It also warns us to be wary of human and computer error, particularly in the form of unusual numbers (Twyman's Law).

We next dicuss the topic of effect sizes. In the instance of conducting an A/B test on groups A and B, we found the difference between revenues was $125. 125 is our effect size in this scenario and shows there is a difference between groups A and B because there is a non-zero effect size. To interpret and make comparisons, we divide the effect size by the standard deviation of a dataset. This is known as Cohen's d.

Now, the textbook covers statistical power -- "the probability that a correctly run A/B test will reject a false null hypothesis" (88). The greater the sample size of a test, the higher its statistical power will be.

The chapter ends by showing the broad array of applications that A/B testing can have, beyond the scope of business and marketing. It also forces us to consider the ethical ramifications of A/B testing and impresses on us the importance of having informed consent for the use of data.
